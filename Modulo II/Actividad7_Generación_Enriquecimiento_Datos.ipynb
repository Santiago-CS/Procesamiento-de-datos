```python
# Desarrolla el apartado 1. Agrega las celdas que necesites.
import os
import numpy as np
import pandas as pd
from IPython.display import display

carpeta_datos = 'Data'

# 1. Cargar penguins.csv conservando el original
pinguinos = pd.read_csv(os.path.join(carpeta_datos, 'penguins.csv'))

# Identificar columnas necesarias y revisar valores faltantes
columnas_necesarias = ['species', 'island', 'body_mass_g', 'flipper_length_mm']
print("=== Valores faltantes en las columnas seleccionadas ===")
display(pinguinos[columnas_necesarias].isna().sum().to_frame(name='faltantes'))

# 2. Crear copia y calcular la nueva variable derivada: Índice Masa-Aleta (g/mm)
pinguinos_enriquecidos = pinguinos.copy()
pinguinos_enriquecidos['indice_masa_aleta'] = (pinguinos_enriquecidos['body_mass_g'] / pinguinos_enriquecidos['flipper_length_mm'])

# 3. Mostrar las columnas originales junto con la nueva en una tabla de 10 filas
print("\n=== Muestra de 10 filas con la nueva variable ===")
columnas_mostrar = ['species', 'island', 'body_mass_g', 'flipper_length_mm', 'indice_masa_aleta']
display(pinguinos_enriquecidos[columnas_mostrar].head(10).round(2))

# 4. Verificación manual de la primera fila (índice 0) y conteo de faltantes
fila_0 = pinguinos_enriquecidos.loc[0]
calculo_manual = fila_0['body_mass_g'] / fila_0['flipper_length_mm']
print(f"\n=== Verificación manual (Fila 0) ===")
print(f"Masa corporal: {fila_0['body_mass_g']} g | Longitud de aleta: {fila_0['flipper_length_mm']} mm")
print(f"Cálculo manual (3750 / 181): {calculo_manual:.4f} g/mm")
print(f"Valor en la columna:         {fila_0['indice_masa_aleta']:.4f} g/mm")
print(f"\nValores faltantes en 'indice_masa_aleta': {pinguinos_enriquecidos['indice_masa_aleta'].isna().sum()}")
```

### Documentación y análisis de `indice_masa_aleta`

* **Pregunta que responde:** ¿Cuántos gramos de masa corporal posee un pingüino por cada milímetro de longitud de su aleta (indicador de robustez o corpulencia relativa)?
* **Fórmula y unidades:** $\text{indice\_masa\_aleta} = \frac{\text{body\_mass\_g}}{\text{flipper\_length\_mm}}$, expresada en gramos por milímetro ($\text{g/mm}$).
* **Verificación y datos faltantes:** Para la fila 0 (*Adelie* en *Torgersen*), se divide $3750\text{ g} / 181\text{ mm} = 20.7182\text{ g/mm}$ ($\approx 20.72\text{ g/mm}$), coincidiendo exactamente con el valor calculado por pandas. La nueva columna tiene **2 valores faltantes**, heredados directamente de los 2 registros que no tienen mediciones en `body_mass_g` ni en `flipper_length_mm`.

**¿Qué aporta la variable?**
Analizar el peso de forma aislada puede ser engañoso porque un pingüino más grande naturalmente pesa más. La variable `indice_masa_aleta` permite comparar la complexión física entre especies controlando por el tamaño de la extremidad: por ejemplo, ayuda a distinguir rápidamente en un reporte si una especie es más pesada en proporción a su tamaño estructural o si su mayor peso se acompaña de aletas proporcionalmente más largas.

## 2. Bike Sharing: enriquecer y comparar registros de alquiler

**¿cómo cambian los resúmenes de demanda cuando modificamos las variables o las reglas de agrupación?**

Carga `hour.csv` y recupera del cuaderno 3.6 las transformaciones necesarias para construir `periodo_dia`, `estacion_nombre` y `clima_nombre`. Consulta la documentación local `bike_sharing_Readme.txt` cuando necesites verificar el significado de una columna.

### 2.1 Modificar los periodos del día

1. Construye la versión original de `demanda_por_periodo`, tal como se trabajó en 3.6, para usarla como referencia.
2. Define una **segunda agrupación horaria**, diferente de la original. Indica los límites, las etiquetas y qué extremos incluye cada intervalo. Los grupos deben cubrir las horas de 0 a 23 sin solapamientos ni horas sin asignar.
3. Genera la nueva tabla con el promedio de alquileres y el número de registros horarios por grupo, conservando la separación entre fin de semana y entre semana.
4. Compara ambas tablas y explica al menos dos cambios observados. Considera que los nuevos grupos pueden tener duraciones distintas.



```python
# Desarrolla el apartado 2.1

# Cargar hour.csv y conservar el original
bicicletas_original = pd.read_csv(os.path.join(carpeta_datos, 'hour.csv'))
bicicletas = bicicletas_original.copy()

# Recuperar transformaciones necesarias del cuaderno de referencia (3.4)
bicicletas['fecha'] = pd.to_datetime(bicicletas['dteday'])
bicicletas['es_fin_de_semana'] = bicicletas['fecha'].dt.dayofweek.isin([5, 6])
bicicletas['humedad_pct'] = bicicletas['hum'] * 100

etiquetas_estacion = {1: 'Invierno', 2: 'Primavera', 3: 'Verano', 4: 'Otoño'}
etiquetas_clima = {
    1: 'Despejado o parcialmente nublado',
    2: 'Neblina o nublado',
    3: 'Nieve o lluvia ligera',
    4: 'Condiciones severas'
}
bicicletas['estacion_nombre'] = bicicletas['season'].map(etiquetas_estacion)
bicicletas['clima_nombre'] = bicicletas['weathersit'].map(etiquetas_clima)

# 1. Versión original de periodo_dia y demanda_por_periodo (bloques uniformes de 6 horas)
bicicletas['periodo_dia'] = pd.cut(
    bicicletas['hr'],
    bins=[0, 6, 12, 18, 24],
    labels=['Madrugada', 'Mañana', 'Tarde', 'Noche'],
    right=False,
    include_lowest=True
)

demanda_por_periodo = (
    bicicletas
    .groupby(['es_fin_de_semana', 'periodo_dia'], as_index=False, observed=True)
    .agg(
        horas_observadas=('cnt', 'size'),
        alquileres_promedio=('cnt', 'mean')
    )
)

# 2. Definir segunda agrupación horaria orientada a horas pico y valle
# Con right=False, cada intervalo es [a, b): incluye el límite izquierdo 'a' y excluye el derecho 'b'
limites_nuevos = [0, 6, 10, 15, 20, 24]
etiquetas_nuevas = [
    'Madrugada (00-05)',
    'Pico Mañana (06-09)',
    'Mediodía Valle (10-14)',
    'Pico Tarde (15-19)',
    'Noche (20-23)'
]

bicicletas['periodo_dia_operativo'] = pd.cut(
    bicicletas['hr'],
    bins=limites_nuevos,
    labels=etiquetas_nuevas,
    right=False,
    include_lowest=True
)

# 3. Generar la nueva tabla con promedio de alquileres y número de registros
demanda_por_periodo_nuevo = (
    bicicletas
    .groupby(['es_fin_de_semana', 'periodo_dia_operativo'], as_index=False, observed=True)
    .agg(
        horas_observadas=('cnt', 'size'),
        alquileres_promedio=('cnt', 'mean')
    )
)

print("=== 1. Tabla Original (demanda_por_periodo) ===")
display(demanda_por_periodo.round(2))

print("\n=== 2. Nueva Tabla con Agrupación Operativa (demanda_por_periodo_nuevo) ===")
display(demanda_por_periodo_nuevo.round(2))
```

### Definición de la nueva agrupación y comparación (Apartado 2.1)

**Regla de la segunda agrupación (`bins=[0, 6, 10, 15, 20, 24]`, `right=False`, `include_lowest=True`):**
Cada intervalo es cerrado por la izquierda y abierto por la derecha $[a, b)$, cubriendo las horas enteras de `0` a `23` sin solapamientos ni valores sin asignar:
* **`Madrugada (00-05)`** $[0, 6)$: incluye las horas `0, 1, 2, 3, 4, 5` (6 horas de duración).
* **`Pico Mañana (06-09)`** $[6, 10)$: incluye las horas `6, 7, 8, 9` (4 horas de duración).
* **`Mediodía Valle (10-14)`** $[10, 15)$: incluye las horas `10, 11, 12, 13, 14` (5 horas de duración).
* **`Pico Tarde (15-19)`** $[15, 20)$: incluye las horas `15, 16, 17, 18, 19` (5 horas de duración).
* **`Noche (20-23)`** $[20, 24)$: incluye las horas `20, 21, 22, 23` (4 horas de duración).

**Cambios observados al comparar ambas tablas:**
1. **Visibilización del pico vespertino entre semana:** En la agrupación original de 6 horas, la `Tarde (12-17)` promediaba horas de actividad moderada con el inicio de la salida laboral, y la `Noche (18-23)` mezclaba la hora pico de las 18:00 con la caída nocturna. Al aislar `Pico Tarde (15-19)`, el promedio entre semana sube notablemente (superando los 350 alquileres por hora), mientras que `Noche (20-23)` baja al quedarse solo con las horas de cierre.
2. **Contraste entre semana vs. fin de semana en la mañana y mediodía:** Entre semana, `Pico Mañana (06-09)` tiene un promedio superior al de `Mediodía Valle (10-14)` por el traslado a oficinas y escuelas; en cambio, durante el fin de semana ocurre exactamente lo opuesto: la demanda se concentra mucho más en `Mediodía Valle (10-14)` y `Pico Tarde (15-19)` que entre las 6 y las 9 de la mañana. Además, como los nuevos intervalos duran 4, 5 o 6 horas, la columna `horas_observadas` cambia proporcionalmente a la duración de cada bloque.

### 2.2 Comparar usuarios casuales y registrados

1. Construye `demanda_por_hora` con los promedios de `casual` y `registered` para cada hora.
2. Calcula la **diferencia absoluta entre ambos promedios**, ordena de mayor a menor y muestra las tres horas con mayor diferencia. Si hay empates en el tercer lugar, indícalos.
3. Interpreta el resultado en 3 a 5 líneas. Aclara que cada fila original representa un registro horario y que el conteo de alquileres no equivale necesariamente a personas únicas.



```python
# Desarrolla el apartado 2.2

# 1. Construir demanda_por_hora con los promedios de casual y registered
demanda_por_hora = (
    bicicletas
    .groupby('hr', as_index=False, observed=True)
    .agg(
        casuales_promedio=('casual', 'mean'),
        registrados_promedio=('registered', 'mean')
    )
)

# 2. Calcular la diferencia absoluta entre ambos promedios y ordenar de mayor a menor
demanda_por_hora['diferencia_absoluta'] = (
    (demanda_por_hora['registrados_promedio'] - demanda_por_hora['casuales_promedio']).abs()
)

demanda_ordenada = demanda_por_hora.sort_values(by='diferencia_absoluta', ascending=False)

print("=== Top 5 horas con mayor diferencia absoluta entre registrados y casuales ===")
display(demanda_ordenada.head(5).round(2))

print("\n=== Las 3 horas con mayor diferencia absoluta ===")
display(demanda_ordenada.head(3).round(2))
```

### Interpretación de diferencias entre usuarios casuales y registrados (Apartado 2.2)

Las tres horas con mayor diferencia absoluta entre el promedio de alquileres registrados y casuales son las **17:00 (`hr=17`)**, las **18:00 (`hr=18`)** y las **8:00 (`hr=8`)**, sin que existan empates en el tercer lugar. Este comportamiento refleja que el uso intensivo del sistema en las horas de entrada (8:00) y salida laboral (17:00–18:00) está dominado por usuarios con membresía (`registered`), mientras que los usuarios ocasionales (`casual`) mantienen un volumen mucho menor en esas franjas. Cabe aclarar que cada fila del dataset original representa un registro acumulado de una hora del calendario y que el conteo de alquileres mide eventos de préstamo de bicicletas, por lo que no equivale necesariamente a personas únicas (un mismo usuario puede realizar múltiples viajes en distintas horas o días).

### 2.4 Proponer una variable adicional

Diseña e implementa **una variable nueva que no se haya creado en 3.6** y que ayude a describir la operación del servicio. Documenta su propósito, fórmula o regla, unidades y una limitación. Muestra un ejemplo y verifica su resultado. Si la fórmula incluye una división, explica qué haces cuando el denominador es cero.


```python
# Desarrolla el apartado 2.4

# Diseñar e implementar: razon_casual_a_registrado = casual / registered
# Si registered == 0, reemplazamos el denominador por NaN con .where() para evitar división entre cero (inf)
bicicletas['razon_casual_a_registrado'] = (
    bicicletas['casual'] / bicicletas['registered'].where(bicicletas['registered'] != 0)
)

# Mostrar una muestra de registros y verificar manualmente la fila 0
columnas_ejemplo = ['dteday', 'hr', 'es_fin_de_semana', 'casual', 'registered', 'cnt', 'razon_casual_a_registrado']
print("=== Muestra de la nueva variable 'razon_casual_a_registrado' ===")
display(bicicletas[columnas_ejemplo].head(8).round(4))

# Verificación manual de la fila 0
f0 = bicicletas.loc[0]
calc_manual_bici = f0['casual'] / f0['registered']
print(f"\n=== Verificación manual (Fila 0) ===")
print(f"Casuales: {f0['casual']} | Registrados: {f0['registered']}")
print(f"Cálculo manual (3 / 13): {calc_manual_bici:.4f}")
print(f"Valor en la columna:     {f0['razon_casual_a_registrado']:.4f}")
print(f"Casos con denominador cero (registered == 0) convertidos a NaN: {(bicicletas['registered'] == 0).sum()}")
```

### Documentación de `razon_casual_a_registrado` (Apartado 2.4)

* **Propósito:** Identificar en qué horas y días la demanda de usuarios ocasionales/turistas (`casual`) se acerca o supera a la de los usuarios con membresía anual (`registered`), ayudando a planear campañas de conversión de membresías o reabastecimiento en zonas recreativas.
* **Fórmula y unidades:** $\text{razon\_casual\_a\_registrado} = \frac{\text{casual}}{\text{registered}}$. Es una razón adimensional (alquileres casuales por cada alquiler registrado; un valor $> 1.0$ indica que en esa hora hubo más viajes casuales que registrados).
* **Manejo de división entre cero:** Cuando en una hora de madrugada no hay ningún usuario registrado (`registered == 0`), la división directa produciría infinito (`inf`) o `NaN` si `casual` también es cero. Mediante `.where(bicicletas['registered'] != 0)`, cualquier cero en el denominador se convierte preventivamente en `NaN` (ocurre en 24 registros horarios), evitando distorsionar promedios posteriores.
* **Verificación:** En la fila 0 (`2011-01-01`, `hr=0`), `casual = 3` y `registered = 13`; al dividir $3 / 13 = 0.2308$, el resultado coincide exactamente con la columna generada.
* **Limitación:** Es muy sensible a denominadores pequeños: por ejemplo, en la madrugada pasar de 1 a 2 usuarios registrados con 3 casuales reduce la razón a la mitad ($3.0$ a $1.5$), aunque el volumen absoluto de bicicletas en uso sea mínimo.
