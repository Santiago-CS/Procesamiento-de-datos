### NumPy

**Ejercicio 1:**

1. Crear un vector con valores del 1 al 120.
2. Crear una función recibe_vector que reciba el vector y, dentro de la función, eleve al cuadrado aquellos números divisibles por 3, y regrese esos números en un nuevo vector.
3. Redimensionar el vector que devuelve la función anterior en una matriz de 20×2.
4. Crear una nueva matriz de 2×10 y multiplicarla por la matriz anterior.
5. Obtener la transpuesta de la matriz resultante.

```python
#Código aquí
import numpy as np

# crear un vector con valores del 1 al 120
# el último número no se incluye, así llega hasta el 120.
vector_original = np.arange(1, 121)

# crear la función recibe_vector
def recibe_vector(vector):
    # Encontramos los números divisibles por 3 (cuyo residuo al dividir entre 3 es 0)
    divisibles_por_3 = vector[vector % 3 == 0]
    
    # Elevamos esos números al cuadrado usando ** 2
    vector_nuevo = divisibles_por_3 ** 2
    
    return vector_nuevo

#llamar a la función y guardar el resultado
vector_procesado = recibe_vector(vector_original)

# Redimensionar a una matriz de 20x2
matriz_20x2 = vector_procesado.reshape(20, 2)

# Crear una nueva matriz de 2x10 y multiplicarla
matriz_2x10 = np.ones((2, 10))

# Multiplicar una matriz de 20x2 por una de 2x10 da como resultado una matriz de 20x10.
# Nuevo descubrimiento para multiplicar estas matrices de velda es con @
matriz_resultante = matriz_20x2 @ matriz_2x10

# Obtener la transpuesta
# La transpuesta voltea la matriz (convierte filas en columnas).
# Nuestra matriz de 20x10 se convertirá en una de 10x20. uso de .T para hacerlo.
matriz_final = matriz_resultante.T

# la forma (tamaño) y el resultado para comprobar
print("Tamaño final:", matriz_final.shape)
print(matriz_final)
```

**Ejercicio 2:**

Se desea calcular el índice de masa corporal (IMC) de jugadores de un equipo de futbol. Los datos proporcionados corresponden a las listas de los nombres, altura y peso, tal como se muestra a continuación

```python
nombre= ['Bonifacio de Cifuentes',
'Ángel Machado Lara',
'Geraldo de Atienza',
'Onofre Posada Grande',
'Samu Cantón-Figuerola',
'Camilo Pol',
'Toni Galvez Lara',
'Lucho Manzano Rivas',
'Abraham Aroca',
'Anselmo Bartolomé Folch',
'Toni Inocencio Rozas Porras',
'Fausto Cortés-Arco',
'Baltasar Carreño Isern']

altura = [180, 215, 210, 210, 188, 176, 209, 200, 210, 188, 176, 209, 200]
peso = [69, 74, 72, 75, 68, 70, 71, 73, 69, 74, 72, 75, 68]
```

El valor de IMC por cada jugador resulta de aplicar la fórmula:

$$IMC = \frac{peso(kg)}{altura(m)^2}$$

**Nota:** Los valores de altura están dados en centímetros por lo que deberá hacer laconversión a metros previo a la aplicación de la fórmula. Adicionalmente, una vez obtenido el IMC se solicita:

> a) Presentar el icm para cada jugador.

> b) Indicar en qué índice se encuentra el icm más alto y a qué jugador le pertenece.

> c) Calcular el promedio de los icm resultantes.

> d) Muestre los nombres de los jugadores en orden de su imc, desde el que tiene menor imc al que tiene mayor imc.

```python
#Código aquí
#la lista vacía para guardar los resultados
imc = []

print("--- a) IMC de cada jugador ---")
for i in range(len(nombre)):
    altura_m = altura[i] / 100
    
    calculo = peso[i] / (altura_m * altura_m) # En lugar de usar potencias, multiplico altura por altura
    imc.append(calculo)
    
    # round() para redondear a 2 decimales sin meter formatos de texto raros
    print(nombre[i], "-", round(calculo, 2))


print("\n--- b) IMC más alto ---")
# buscar el mayor manualmente usando variables de apoyo
indice_mayor = 0
imc_mayor = 0

for i in range(len(imc)):
    # Si el IMC que estamos revisando ahorita es mayor al que teníamos guardado...
    if imc[i] > imc_mayor:
        imc_mayor = imc[i] # ... lo guardamos como el nuevo rey
        indice_mayor = i   # ... y guardamos en qué posición estaba

print("El mayor es de", nombre[indice_mayor], "con un IMC de", round(imc_mayor, 2))


print("\n--- c) Promedio ---")
# la suma "a pie"
suma_total = 0
for i in range(len(imc)):
    suma_total = suma_total + imc[i]

promedio = suma_total / len(imc)
print("El promedio es:", round(promedio, 2))

 
print("\n--- d) Jugadores ordenados por su IMC ---")
# En lugar de zip(), armamos una lista que contenga sub-listas de "parejitas" [IMC, Nombre]
lista_parejas = []
for i in range(len(nombre)):
    pareja = [imc[i], nombre[i]]
    lista_parejas.append(pareja)

lista_parejas.sort()

# imprimir la lista de parejitas ya ordenada
for i in range(len(lista_parejas)):
    # lista_parejas[i][0] es el número del IMC
    # lista_parejas[i][1] es el texto del Nombre
    valor_imc = lista_parejas[i][0]
    jugador = lista_parejas[i][1]
    
    print(jugador, "-", round(valor_imc, 2))
```

### Pandas

**Ejercicio 3:**

Usando el archivo `housing_prices.csv`:

1. Carga el archivo en un `DataFrame` con `pd.read_csv`.
2. Muestra las primeras 5 filas y las estadísticas descriptivas.
3. Crea una columna `precio_por_pie` = `price` / `size`.
4. Filtra y muestra solo las casas con más de 3 recámaras.
5. Calcula el precio promedio agrupando por número de recámaras (`groupby`).
6. Grafica un histograma de la columna `price` con `df['price'].plot(kind='hist')`.


```python
#Código aquí
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/Santiago-CS/Procesamiento-de-datos/refs/heads/main/Inicial/housing_prices.csv"

df = pd.read_csv(url)
print("Datos cargados exitosamente en DataFrame de pandas:")

print(df.head()) 

print("\n--- Estadísticas descriptivas ---")
# describe() saca automáticamente conteo, promedio, mínimo, máximo, etc., de todo.
print(df.describe()) 

#Crear la columna precio_por_pie = price / size
df['precio_por_pie'] = df['price'] / df['size']

# Filtrar y mostrar solo las casas con más de 3 recámaras
casas_grandes = df[df['n_bedrooms'] > 3]

print("\n--- Casas con más de 3 recámaras ---")
# Imprimo solo las primeras 5 de este filtro
print(casas_grandes.head()) 

# Calcular el precio promedio agrupando por número de recámaras
# groupby() junta las filas que tienen el mismo número de recámaras, y mean() saca el promedio de sus precios
precio_promedio = df.groupby('n_bedrooms')['price'].mean()

print("\n--- Precio promedio por número de recámaras ---")
print(precio_promedio)

# Graficando el histograma de la columna price
df['price'].plot(kind='hist', title='Distribución de Precios', color='skyblue', edgecolor='black')
plt.show()
```
