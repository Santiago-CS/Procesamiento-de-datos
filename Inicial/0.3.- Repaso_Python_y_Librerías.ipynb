# 1. Estructuras de datos en Python

Una **estructura de datos** es una forma de organizar y almacenar varios valores dentro de una sola variable. Python trae cuatro estructuras integradas que son la base de todo el análisis de datos:

```mermaid
flowchart TD
    ED["Estructuras de datos<br/>integradas en Python"]
    ED --> L["list · lista<br/>[ ]"]
    ED --> T["tuple · tupla<br/>( )"]
    ED --> D["dict · diccionario<br/>{ clave: valor }"]
    ED --> S["set · conjunto<br/>{ }"]
    L --> L1["Ordenada · Mutable<br/>Permite duplicados"]
    T --> T1["Ordenada · Inmutable<br/>Permite duplicados"]
    D --> D1["Pares clave-valor<br/>Claves unicas"]
    S --> S1["Sin orden · Mutable<br/>Sin duplicados"]
```

| Estructura | Sintaxis | ¿Ordenada? | ¿Mutable? | ¿Duplicados? | Ejemplo de uso |
|-----------|----------|:----------:|:---------:|:------------:|----------------|
| **Lista** (`list`) | `[ ]` | ✅ | ✅ | ✅ | Colección de valores que cambia |
| **Tupla** (`tuple`) | `( )` | ✅ | ❌ | ✅ | Datos fijos (coordenadas, RGB) |
| **Diccionario** (`dict`) | `{k: v}` | ✅* | ✅ | Claves no | Buscar por clave (nombre → dato) |
| **Conjunto** (`set`) | `{ }` | ❌ | ✅ | ❌ | Elementos únicos, operaciones de conjuntos |

<sub>*Desde Python 3.7 los diccionarios conservan el orden de inserción.</sub>

## 3.1 Diccionarios (`dict`)

Un **diccionario** guarda pares **clave → valor**. En vez de acceder por posición (como en las listas), accedemos por su **clave**. Las claves deben ser **únicas**.

```mermaid
flowchart LR
    subgraph D["escuela = { }"]
      direction TB
      k1["'Nombre'"] --> v1["'ITESO'"]
      k2["'Ubicacion'"] --> v2["'Av. mmad'"]
      k3["'Horarios'"] --> v3["'9am - 6pm'"]
    end
```

| Operación | Sintaxis | Devuelve |
|-----------|----------|----------|
| Acceder a un valor | `d['clave']` | El valor asociado |
| Ver todas las claves | `d.keys()` | Lista de claves |
| Ver todos los valores | `d.values()` | Lista de valores |
| Ver pares clave-valor | `d.items()` | Pares `(clave, valor)` |
| Agregar / actualizar | `d['nueva'] = 10` | — |

>  La expresión `{k: v for (k, v) in zip(claves, valores)}` es una **comprensión de diccionario**: construye un `dict` a partir de dos listas.

## 3.3 Conjuntos (`set`)

Un **conjunto** es una colección **desordenada** de elementos **únicos** (sin duplicados). Es ideal para eliminar repetidos y hacer operaciones de teoría de conjuntos (unión, intersección, diferencia).

```mermaid
flowchart LR
    A["Lista con duplicados<br/>[1, 2, 2, 3, 3, 3]"] -->|set| B["Conjunto<br/>{1, 2, 3}"]
```

| Operación | Símbolo | Método | Significado |
|-----------|:-------:|--------|-------------|
| Unión | `A | B` | `A.union(B)` | Elementos en A **o** B |
| Intersección | `A & B` | `A.intersection(B)` | Elementos en A **y** B |
| Diferencia | `A - B` | `A.difference(B)` | En A pero **no** en B |

>  Se crean con `{ }` o con `set(...)`. Un `{ }` vacío es un **diccionario**, no un conjunto: para un conjunto vacío usa `set()`.


````python
# Un conjunto elimina automáticamente los duplicados
numeros = [1, 2, 2, 3, 3, 3, 4]
conjunto = set(numeros)
conjunto
````

````python
# Operaciones entre conjuntos
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print('Unión:', A | B)
print('Intersección:', A & B)
print('Diferencia A - B:', A - B)
````

La función principal es `plot()`, que toma un número variable de argumentos y grafica de forma tan simple como escribiendo `plot(x, y)`.

```mermaid
flowchart LR
    A["Datos x, y"] --> B["plt.plot(x, y)"]
    B --> C["Titulo, ejes,<br/>leyenda, grid"]
    C --> D["plt.show()"]
```

| Función | Para qué sirve |
|---------|----------------|

# 5. Definiendo funciones

Las funciones por defecto de Python son de mucha utilidad. Sin embargo, pronto nos daremos cuenta de que es aún más útil **definir nuestras propias funciones** para reutilizarlas cada vez que las necesitemos.

Una función recibe **entradas** (argumentos), ejecuta un bloque de código y opcionalmente **devuelve** un resultado con `return`.

```mermaid
flowchart LR
    A["Entradas<br/>argumentos"] --> B["def funcion(...):<br/>bloque de codigo"]
    B --> C["return<br/>resultado"]
```

>  **Anatomía:** `def nombre(args):` inicia la función; el código **indentado** es su cuerpo; `return` termina la función y entrega un valor. El texto entre `''' '''` al inicio es el *docstring* (documentación que muestra `help()`).

Por ejemplo, creemos una función que dados tres números, devuelva la mínima diferencia absoluta entre ellos.


````python
# Explicar acá la forma de definir una función
def diferencia_minima(a,b,c):
    '''
    Esta función regresa la diferencia mínima absoluta de tres números
    '''
    diff1 = abs(a-b)
    diff2 = abs(a-c)
    diff3 = abs(b-c)
    
    return min(diff1,diff2,diff3)
    
````

Las funciones comienzan con la palabra clave `def`, y el código indentado luego de los dos puntos `:` se corre cuando la función es llamada.

`return` es otra parablra clave que sólo se asocia con funciones. Cuando Python se encuentra un `return`, termina la función inmediatamente y devuelve el valor que hay seguido del `return`.

¿Qué hace específicamente la función que escribimos?

````python
# Ejemplo: llamar la función unas 3 veces
diferencia_minima(1,2,3)
````

````python
diferencia_minima(-1,2,3)
````

````python
diferencia_minima(-2.0,2.5,3.3)
````

````python
help(diferencia_minima)
````

Intentemos llamar `help` sobre la función

Bueno, Python tampoco es tan listo como para leer código y entregar una buena descripción de la función. Esto es trabajo del diseñador de la función: incluir la documentación.

¿Cómo se hace? (Recordar añadir un ejemplo)

````python
# Copiar y pegar la función, pero esta vez, incluir documentación de la misma
def diferencia_minima(a,b,c):
    '''
    Esta función regresa la diferencia mínima absoluta de tres números, recibe 3 argumentos a,b,c
    y devuelve la dif mínima.
    Ejemplo:
    diferencia_minima(-2.0,2.5,3.3)
    out: 0.7999999999999998
    '''
    diff1 = abs(a-b)
    diff2 = abs(a-c)
    diff3 = abs(b-c)
    
    return min(diff1,diff2,diff3)
    
````

````python
# Volver a llamar el help
help(diferencia_minima)
````

Muy bien. Ahora, podemos observar que podemos llamar esta función sobre diferentes números, incluso de diferentes tipos:

- Si todos son enteros, entonces nos retornará un entero.
- Si hay algún float, nos retornará un float.

````python
# Todos enteros
diferencia_minima(1,2,3)
````

````python
# Uno o más floats
diferencia_minima(-1.0,2.0,3.0)
````

````python
# String: TypeError
diferencia_minima('q','a','c')
````

## 6.1 Series

Una **`Series`** es una columna de datos con un **índice** (etiquetas) a la izquierda. Se puede crear a partir de una lista o de un diccionario.


````python
# Reutilizamos los datos de temperatura de la sección de diccionarios
dias = ['Lun', 'Mar', 'Mier', 'Jue', 'Vie', 'Sab', 'Dom']
tmp = [28.0, 28.3, 27.0, 27.5, 28.6, 29.0, 30.0]

temperaturas = pd.Series(tmp, index=dias, name='Temperatura')
temperaturas
````

### Agrupar y resumir (`groupby`)

Una operación muy potente es **agrupar** filas que comparten un valor y calcular un resumen por grupo (promedio, conteo, suma...). Aquí agrupamos por número de recámaras.
