## 📖 Descripción de la Asignatura

En la actualidad a cada minuto se capturan miles o millones de datos sobre la medición de fenómenos naturales a través de diversos medios, experimentos de laboratorio, en las búsquedas, tendencias e intereses de usuarios de internet, etc. Los conjuntos de datos (dataset) que se generan pueden contener datos espurios, repetidos, faltantes, registros incorrectos, entre otros, por lo cual es importante que para convertir los datos en información se realice un preprocesamiento como lo son la limpieza de datos, la normalización y la reducción de dimensiones.

El objetivo es construir y preparar un conjunto de datos (dataset) cuyo origen puede provenir de diferentes fuentes, mediante el uso de herramientas computacionales.
