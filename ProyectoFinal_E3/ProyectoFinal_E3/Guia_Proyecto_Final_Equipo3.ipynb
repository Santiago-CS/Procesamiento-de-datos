## Etapa 2. Fuente y extracción de datos: 

Elijan el método de lectura de acuerdo con el formato disponible, por ejemplo CSV, Excel, JSON, XML, texto, Parquet, HTML o datos geográficos e imágenes cuando sean pertinentes.
```python
import xml.etree.ElementTree as ET
import pandas as pd

ruta = "Data/Cuisine_rating.xml"
root = ET.parse(ruta).getroot()

filas = []
for rec in root.findall("record"):
    cliente = {}
    for f in rec.findall("field"):
        columna = f.attrib.get("name", f.tag)
        valor = f.text
        cliente[columna] = valor
    filas.append(cliente)

df = pd.DataFrame(filas)

df_original = df.copy(deep=True)
```

### 3.4 Distribuciones, frecuencias categóricas y detección de atípicos

Se calculan las medidas de tendencia central y dispersión de las variables numéricas observadas, junto con los límites de Tukey (`Q1 - 1.5 * IQR` y `Q3 + 1.5 * IQR`) para identificar posibles valores atípicos, así como los conteos de las variables categóricas.

```python
vars_num_analisis = ["YOB", "Budget", "Food Rating", "Service Rating", "Overall Rating"]

resumen_num = []
for col in vars_num_analisis:
    s = df_exp[col].dropna()
    q1, med, q3 = s.quantile([0.25, 0.50, 0.75])
    iqr = q3 - q1
    lim_inf = q1 - 1.5 * iqr
    lim_sup = q3 + 1.5 * iqr
    atipicos = ((s < lim_inf) | (s > lim_sup)).sum()
    resumen_num.append({
        "Variable": col,
        "Media": round(s.mean(), 2),
        "Mediana": round(med, 2),
        "Desv. Est.": round(s.std(), 2),
        "Mín": s.min(),
        "Q1": q1,
        "Q3": q3,
        "Máx": s.max(),
        "Lím. Inf. IQR": round(lim_inf, 2),
        "Lím. Sup. IQR": round(lim_sup, 2),
        "Atípicos": atipicos
    })

display(pd.DataFrame(resumen_num))

fig, axes = plt.subplots(2, 5, figsize=(16, 6.5))
for i, col in enumerate(vars_num_analisis):
    s = df_exp[col].dropna()
    axes[0, i].hist(s, bins=10, color="#6c8ebf", edgecolor="black")
    axes[0, i].set_title(f"Histograma: {col}", fontsize=9)
    axes[0, i].set_ylabel("Frecuencia" if i == 0 else "")
    axes[1, i].boxplot(s, vert=True, patch_artist=True,
                       boxprops=dict(facecolor="#d5e8d4", color="black"),
                       medianprops=dict(color="black", linewidth=2))
    axes[1, i].set_title(f"Boxplot: {col}", fontsize=9)
    axes[1, i].set_xticks([])
plt.tight_layout()
plt.show()

print("Frecuencias de variables categóricas:")
for col in ["Cuisines", "Gender", "Marital Status", "Activity", "Alcohol", "Smoker", "Often A S"]:
    print(f"\n--- {col} ---")
    print(df_exp[col].value_counts(dropna=False))
```

### 4.1 Copia de trabajo, tipos físicos y estandarización de `Location`

Se trabaja sobre una copia profunda `df_limpio = df_original.copy(deep=True)`. Primero se corrige el nombre `'Alcohol '` a `'Alcohol'`, se convierten las columnas numéricas a su tipo correspondiente y se unifica el formato de `Location` a `'Ciudad, NY'` (corrigiendo minúsculas en `'Central Park,ny'`, el error `'Market City, MY'` por `'Market City, NY'` y los espacios tras la coma).

```python
df_limpio = df_original.copy(deep=True)
df_limpio = df_limpio.rename(columns={"Alcohol ": "Alcohol"})

for col in ["User ID", "Area code", "YOB", "Budget", "Food Rating", "Service Rating", "Overall Rating"]:
    df_limpio[col] = pd.to_numeric(df_limpio[col], errors="coerce")

loc_antes = df_limpio["Location"].copy()
# Normalizar espacios alrededor de la coma y corregir errores conocidos
mask_loc = df_limpio["Location"].notna()
df_limpio.loc[mask_loc, "Location"] = (
    df_limpio.loc[mask_loc, "Location"]
    .str.strip()
    .str.replace(r"\s*,\s*", ", ", regex=True)
    .str.replace(", ny", ", NY", regex=False)
    .str.replace(", MY", ", NY", regex=False)
)

cambios_loc = (loc_antes.fillna("") != df_limpio["Location"].fillna("")).sum()
print(f"Registros modificados en Location al estandarizar formato y corregir estado: {cambios_loc}")
print(df_limpio["Location"].value_counts(dropna=False))
```

### 4.2 Tratamiento de valores faltantes

Se aplica una estrategia en tres niveles para evitar distorsionar las distribuciones cuando existe información exacta disponible:
1. **Reconstrucción determinista de calificaciones:** En los 25 registros donde falta `Food Rating` (11) o `Service Rating` (14), la otra calificación y `Overall Rating` están presentes. Usando `Overall Rating = (Food Rating + Service Rating) / 2`, se recupera el valor exacto (`2 * Overall Rating - otra_calificacion`).
2. **Reconstrucción lógica en hábitos (`Often A S`, `Alcohol`, `Smoker`):** Si `Often A S == 'Yes'`, ambos hábitos deben ser `'Often'`. Si `Often A S` es nulo pero `Alcohol` o `Smoker` es `'Never'` o `'Socially'`, entonces `Often A S` es `'No'`.
3. **Imputación de faltantes restantes:**
   - Para `YOB`, `Budget` y `Area code` se comparan **mediana** y **`KNNImputer(n_neighbors=5)`** (estandarizando antes de calcular distancias y regresando a la escala original). Dado que el porcentaje de faltantes es bajo (4.5% a 7.0%) y las variables son discretas/enteras, se evalúa cuál preserva mejor la mediana y desviación estándar originales.
   - Para las variables categóricas restantes (`Location`, `Gender`, `Marital Status`, `Activity`, `Cuisines`, `Alcohol`, `Smoker`, `Often A S`) se imputa con la **moda** de cada columna, ya que la ausencia es menor al 6% y no se busca crear categorías artificiales con pocos casos.

```python
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler

# 1. Reconstrucción determinista de Food Rating y Service Rating
mask_food_na = df_limpio["Food Rating"].isna() & df_limpio["Service Rating"].notna()
mask_serv_na = df_limpio["Service Rating"].isna() & df_limpio["Food Rating"].notna()

df_limpio.loc[mask_food_na, "Food Rating"] = 2 * df_limpio.loc[mask_food_na, "Overall Rating"] - df_limpio.loc[mask_food_na, "Service Rating"]
df_limpio.loc[mask_serv_na, "Service Rating"] = 2 * df_limpio.loc[mask_serv_na, "Overall Rating"] - df_limpio.loc[mask_serv_na, "Food Rating"]

print(f"Food Rating reconstruidos exactamente: {mask_food_na.sum()}")
print(f"Service Rating reconstruidos exactamente: {mask_serv_na.sum()}")

# 2. Reconstrucción lógica de Often A S, Alcohol y Smoker
mask_yes_alc = (df_limpio["Often A S"] == "Yes") & df_limpio["Alcohol"].isna()
mask_yes_smk = (df_limpio["Often A S"] == "Yes") & df_limpio["Smoker"].isna()
df_limpio.loc[mask_yes_alc, "Alcohol"] = "Often"
df_limpio.loc[mask_yes_smk, "Smoker"] = "Often"

mask_oas_no = df_limpio["Often A S"].isna() & (
    df_limpio["Alcohol"].isin(["Never", "Socially"]) | df_limpio["Smoker"].isin(["Never", "Socially"])
)
df_limpio.loc[mask_oas_no, "Often A S"] = "No"
print(f"Alcohol/Smoker deducidos desde Often A S='Yes': {mask_yes_alc.sum() + mask_yes_smk.sum()}")
print(f"Often A S deducidos como 'No': {mask_oas_no.sum()}")

# 3. Comparación Mediana vs KNNImputer(n_neighbors=5) para YOB, Budget y Area code
cols_imp_num = ["Area code", "YOB", "Budget"]
df_mediana = df_limpio[cols_imp_num].copy()
for c in cols_imp_num:
    df_mediana[c] = df_mediana[c].fillna(df_mediana[c].median())

scaler_imp = StandardScaler()
datos_std = scaler_imp.fit_transform(df_limpio[cols_imp_num])
knn = KNNImputer(n_neighbors=5)
datos_knn_std = knn.fit_transform(datos_std)
df_knn = pd.DataFrame(scaler_imp.inverse_transform(datos_knn_std), columns=cols_imp_num, index=df_limpio.index)

comp_imp = []
for c in cols_imp_num:
    obs = df_limpio[c].dropna()
    comp_imp.append({
        "Variable": c,
        "Mediana_Orig": round(obs.median(), 2),
        "Std_Orig": round(obs.std(), 2),
        "Mediana_Med": round(df_mediana[c].median(), 2),
        "Std_Med": round(df_mediana[c].std(), 2),
        "Mediana_KNN": round(df_knn[c].median(), 2),
        "Std_KNN": round(df_knn[c].std(), 2),
    })
display(pd.DataFrame(comp_imp))
```

Ambos métodos mantienen prácticamente idéntica la mediana, y `Mediana` evita introducir valores decimales artificiales en variables enteras como `YOB`, `Budget` y `Area code` (mientras que KNN reduce ligeramente más la desviación en `YOB` y `Budget` y genera promedios fraccionarios). Por ello, se adopta la **imputación por mediana** para estas tres columnas numéricas y la **imputación por moda** para las categóricas restantes.

```python
# Aplicar mediana a numéricas restantes
for c in cols_imp_num:
    val_med = df_limpio[c].median()
    n_imp = df_limpio[c].isna().sum()
    df_limpio[c] = df_limpio[c].fillna(val_med)
    print(f"Imputados en {c}: {n_imp} celdas con mediana = {val_med}")

# Aplicar moda a categóricas restantes
cols_imp_cat = ["Location", "Gender", "Marital Status", "Activity", "Cuisines", "Alcohol", "Smoker", "Often A S"]
imp_cat = SimpleImputer(strategy="most_frequent")
for c in cols_imp_cat:
    n_imp = df_limpio[c].isna().sum()
    if n_imp > 0:
        moda = df_limpio[c].mode().iloc[0]
        df_limpio[c] = df_limpio[c].fillna(moda)
        print(f"Imputados en {c}: {n_imp} celdas con moda = '{moda}'")

print("\nTotal de faltantes después de la limpieza:", df_limpio.isnull().sum().sum())
```

### 5.2 Selección por correlación

Mostrar una matriz o mapa de calor para las características relevantes. Elegir Pearson para asociación lineal o Spearman para asociación monótona por rangos, y justificarlo. Si filtran pares altamente correlacionados, declarar un umbral (por ejemplo, 0.85) y una regla reproducible para decidir cuál conservar. Reportar las columnas retiradas y explicar el criterio.

La correlación mide asociación, no causalidad ni independencia. Correlación baja no descarta dependencia no lineal, un umbral por pares tampoco encuentra redundancia multivariable. 

Se elige la **correlación de Spearman** porque varias variables relevantes (`Budget`, `Food Rating`, `Service Rating`, `Overall Rating`, `Alcohol_ord`, `Smoker_ord`) son ordinales o discretas en escalas de 1 a 5 o 0 a 2, y Spearman evalúa asociaciones monótonas por rangos sin exigir normalidad ni linealidad estricta. Se fija un umbral de redundancia de **$|r| \ge 0.85$**; en caso de superar ese umbral, se conserva la variable más interpretable o base y se retira la redundante.

```python
cols_corr = ["YOB", "Edad", "Budget", "Food Rating", "Service Rating", "Overall Rating", "Alcohol_ord", "Smoker_ord"]
corr_spearman = df_limpio[cols_corr].corr(method="spearman")

plt.figure(figsize=(8.5, 6.5))
sns.heatmap(corr_spearman, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, linewidths=0.5)
plt.title("Matriz de correlación de Spearman (variables numéricas y ordinales)")
plt.tight_layout()
plt.show()

# Detectar pares con |r| >= 0.85
umbral = 0.85
pares_altos = []
for i in range(len(cols_corr)):
    for j in range(i + 1, len(cols_corr)):
        r = corr_spearman.iloc[i, j]
        if abs(r) >= umbral:
            pares_altos.append((cols_corr[i], cols_corr[j], round(r, 3)))

print("Pares con |r| >= 0.85:", pares_altos)
```

El único par que supera el umbral de `0.85` es `YOB` con `Edad` (`r = -1.00`), ya que `Edad = 2026 - YOB` es una transformación lineal exacta. Se retira `YOB` y se conserva `Edad` por su interpretación directa. Además, `Overall Rating` mantiene una correlación alta con `Food Rating` (`0.693`) y `Service Rating` (`0.764`) porque es su promedio aritmético exacto; por esta razón, para evitar redundancia estructural en PCA, se utilizan `Food Rating` y `Service Rating` por separado en lugar de incluir `Overall Rating`.

### 5.5 Análisis de componentes principales (PCA)

PCA es no supervisado: encuentra combinaciones que conservan varianza, sin consultar la etiqueta. Estandarizar las variables cuando sus escalas/unidades sean distintas y justificar el escalado cuando deban conservarse magnitudes originales. El ajuste y el escalador se aprenden solo en entrenamiento.

Mostrar varianza explicada individual y acumulada, justificar cuántos componentes se conservan. Interpretar los pesos principales y, si ayuda, graficar las primeras componentes. 

El **PCA** sí es pertinente porque es un método **no supervisado** que permite resumir la variabilidad conjunta de las variables numéricas y ordinales del cliente (`Edad`, `Budget`, `Food Rating`, `Service Rating`, `Alcohol_ord`, `Smoker_ord`). Dado que `Edad` está en años y el resto en escalas de 1–5 o 0–2, se estandarizan todas las variables con `StandardScaler` antes de ajustar el PCA.

```python
from sklearn.decomposition import PCA

cols_pca = ["Edad", "Budget", "Food Rating", "Service Rating", "Alcohol_ord", "Smoker_ord"]
X_pca_std = StandardScaler().fit_transform(df_limpio[cols_pca])

pca = PCA()
componentes = pca.fit_transform(X_pca_std)

var_ind = pca.explained_variance_ratio_
var_acum = np.cumsum(var_ind)

tabla_pca = pd.DataFrame({
    "Componente": [f"PC{i+1}" for i in range(len(cols_pca))],
    "Varianza explicada (%)": (var_ind * 100).round(2),
    "Varianza acumulada (%)": (var_acum * 100).round(2)
})
display(tabla_pca)

loadings = pd.DataFrame(
    pca.components_[:3].T,
    index=cols_pca,
    columns=["PC1", "PC2", "PC3"]
).round(3)
print("Pesos (loadings) de las primeras 3 componentes:")
display(loadings)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))

# Gráfica de varianza explicada
x_pc = np.arange(1, len(cols_pca) + 1)
axes[0].bar(x_pc, var_ind * 100, color="#6c8ebf", edgecolor="black", label="Individual")
axes[0].plot(x_pc, var_acum * 100, color="black", marker="o", linewidth=2, label="Acumulada")
axes[0].axhline(70, color="gray", linestyle="--", linewidth=1, label="Referencia 70%")
axes[0].set_title("Varianza explicada por componente principal")
axes[0].set_xlabel("Componente")
axes[0].set_ylabel("Porcentaje de varianza (%)")
axes[0].set_xticks(x_pc)
axes[0].set_xticklabels([f"PC{i}" for i in x_pc])
axes[0].legend(fontsize=8)

# Proyección 2D (PC1 vs PC2) diferenciando por nivel de presupuesto con marcadores y tono
marcadores = {1: "v", 2: "s", 3: "o", 4: "^", 5: "D"}
tonos = {1: "#d9d9d9", 2: "#bdbdbd", 3: "#969696", 4: "#636363", 5: "#252525"}
for b in sorted(df_limpio["Budget"].unique()):
    mask_b = df_limpio["Budget"] == b
    axes[1].scatter(componentes[mask_b, 0], componentes[mask_b, 1],
                    marker=marcadores[int(b)], color=tonos[int(b)], edgecolor="black",
                    s=45, alpha=0.85, label=f"Budget {int(b)}")
axes[1].axhline(0, color="gray", linestyle=":", linewidth=0.8)
axes[1].axvline(0, color="gray", linestyle=":", linewidth=0.8)
axes[1].set_title("Proyección PCA (PC1 vs PC2) por nivel de Budget")
axes[1].set_xlabel(f"PC1 ({var_ind[0]*100:.1f}% var.)")
axes[1].set_ylabel(f"PC2 ({var_ind[1]*100:.1f}% var.)")
axes[1].legend(title="Budget", fontsize=8, ncol=2)

plt.tight_layout()
plt.show()
```

**Interpretación del PCA:**
- Las primeras cuatro componentes (`PC1` a `PC4`) acumulan el **76.98%** de la varianza total (`PC1` + `PC2` explican el **45.82%**), lo que indica que las variables no son altamente colineales entre sí y cada dimensión aporta información propia.
- En los pesos (`loadings`), `PC1` combina positivamente los hábitos de consumo (`Smoker_ord`: `0.605`, `Alcohol_ord`: `0.374`) y las calificaciones (`Food Rating`: `0.489`, `Service Rating`: `0.433`), mientras que `PC2` se asocia principalmente con `Budget` (`0.655`), `Food Rating` (`0.493`) y `Edad` (`0.444`).

**(15 min de presentación y 5 min de preguntas):** Debe comunicar el problema, los datos asignados, el método de procesamiento, los hallazgos más importantes, las limitaciones y las conclusiones. Incluyan figuras legibles, títulos y una explicación breve de qué evidencia aporta cada una. Todas las personas del equipo deben poder explicar el trabajo y responder preguntas.

### 2. Reporte digital del trabajo realizado

El **reporte** se entregará como documento digital elaborado en computadora; **no se aceptan fotografías de hojas o cuadernos**. Como mínimo, debe incluir:

- Nombre y apellidos de todas las personas integrantes.
- Descripción del problema, pregunta, objetivo y solución propuesta.
- Fuente, liga del dataset, descripción de datos y diccionario de variables.
- **Proceso realizado (Etapa 1-5)**: extracción, limpieza, decisiones de transformación y métodos utilizados.
- Resultados con figuras/tablas y su interpretación.
- **Un capítulo que describa la solución que se propuso para el problema.**
- **El código generado por el equipo para realizar el proyecto.** Puede organizarse en bloques legibles en el reporte y complementarse con el notebook entregado por separado, debe quedar claro qué código produjo los resultados.
- **Una sección de conclusiones.**
- Limitaciones, referencias y explicación de las decisiones relevantes.

**Impresión:** no es necesario imprimir a color. Diseñen las figuras para que se distingan en escala de grises, usen tipos de línea, marcadores, etiquetas o patrones además del color.

**Estructura sugerida del reporte:** 
- portada e integrantes
- resumen
- contexto de los datos y fuente
- Metodología/proceso
- Capítulo de solución propuesta (Etapas 1- 5)
- Resultados y figuras
- Limitaciones
- Conclusiones, referencias
- Apéndice de código si se requiere.

### 3. Código en Jupyter notebook

El notebook debe documentar el proceso completo: código, análisis, comentarios, tablas y gráficas. Debe ejecutarse en orden y ser comprensible para alguien que no participó en su elaboración.

- Incluir nombres completos, equipo, pregunta y liga asignada.
- Mantener comentarios y texto Markdown que expliquen qué se hace y por qué.
- No omitir celdas necesarias ni depender de variables creadas manualmente en otra sesión.
- Usar rutas relativas, indicar formato y fuentes de datos y no incluir credenciales privadas.
- Mostrar las figuras y salidas que se discuten en el reporte.
- Registrar decisiones, parámetros y semilla aleatoria cuando aplique.
- Ejecutar desde un kernel limpio antes de entregar y revisar que no haya errores.
- Mantener la versión original de los datos; no guardar archivos derivados sobre el original sin indicarlo.
